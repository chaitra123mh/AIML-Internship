```python
import pandas as pd

data = {
    "Name":["Amit","Priya","Rahul","Sneha","Ravi","Neha"],
    "State":["Karnataka","Tamil Nadu","Karnataka","Kerala","Maharashtra","Karnataka"],
    "Salary":[50000,60000,45000,70000,65000,52000],
    "Department":["IT","HR","IT","Finance","HR","IT"]
}

df = pd.DataFrame(data)

df.to_csv("indian_employees.csv",index=False)

df
```

```python
df=pd.read_csv("indian_employees.csv")

df
```

```python
# Filter employees whose salary is greater than 50000

filtered=df[df["Salary"]>50000]

print(filtered)
```

```python
# Group employees by state and calculate average salary

group=df.groupby("State")["Salary"].mean()

print(group)
```

```python
department_info=pd.DataFrame({

"Department":["IT","HR","Finance"],

"Manager":["Kiran","Anita","Suresh"]

})

merged=pd.merge(df,department_info,on="Department")

print(merged)
```

```python
pivot=df.pivot_table(

values="Salary",

index="State",

columns="Department",

aggfunc="mean"

)

print(pivot)
```

```python
df.to_csv("cleaned_data.csv",index=False)
```
