```python
import numpy as np
import pandas as pd

# Create sample data
data = {
    "Name": ["Asha", "Rahul", "Priya", "Rohan", "Sneha"],
    "Math": [85, 78, 92, 74, 88],
    "Science": [90, 82, 95, 70, 91]
}

# Create DataFrame
df = pd.DataFrame(data)

# Save to CSV
df.to_csv("students.csv", index=False)

# Read CSV
df = pd.read_csv("students.csv")

print("Student Data:")
print(df)

# Convert columns to NumPy arrays
math_marks = df["Math"].to_numpy()
science_marks = df["Science"].to_numpy()

# Mean
print("\nMean of Math Marks:", np.mean(math_marks))

# Standard Deviation
print("Standard Deviation:", np.std(math_marks))

# Correlation
print("\nCorrelation Matrix:")
print(np.corrcoef(math_marks, science_marks))
```
